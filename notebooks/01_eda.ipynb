# 01 — Exploratory Data Analysis

All data is synthetic. Ledger: 4,492 AR/AP transactions across WLW-US / WLW-EU / WLW-ASIA. Inventory: 150 watch SKUs. Reconciliation: daily GL vs subledger for 3 entities x 3 subledgers.

```python
import pandas as pd
led = pd.read_csv('data/processed/ledger.csv', parse_dates=['txn_date'])
inv = pd.read_csv('data/processed/inventory_monthly.csv', parse_dates=['month_end'])
rec = pd.read_csv('data/processed/recon_daily.csv', parse_dates=['recon_date'])
print(led.shape, inv.shape, rec.shape)
```

```python
led.groupby(['entity', 'txn_type']).agg(n=('txn_id', 'count'),
    total_usd=('amount_usd', 'sum'), median_usd=('amount_usd', 'median'))
```

```python
led['is_anomaly'].mean(), led.groupby('anomaly_type').size()
```

```python
inv.groupby('month_end')['total_value_usd'].sum().plot(title='Inventory value by month')
```

```python
rec.groupby('break_type').size()
```
