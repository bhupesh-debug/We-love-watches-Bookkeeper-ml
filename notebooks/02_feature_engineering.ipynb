# 02 — Feature Engineering

Counterparty statistics are fitted on the TRAIN period only (no leakage). Key features: log amount, counterparty z-score, same-amount-in-trailing-7-days (the duplicate check), days since last transaction with the same counterparty, weekend flag, supporting-document flag, round-number flag, FX deviation, overdue days.

```python
import sys; sys.path.insert(0, 'src')
import pandas as pd, feature_engineering as fe
led = pd.read_csv('data/processed/ledger.csv', parse_dates=['txn_date', 'due_date', 'paid_date'])
train = led[led['txn_date'] < '2025-11-01']
stats, gmed, gstd = fe.fit_counterparty_stats(train)
print(stats.head())
f = fe.build_ledger_features(train, stats, gmed, gstd)
print(f.shape)
print(f.columns.tolist())
```
